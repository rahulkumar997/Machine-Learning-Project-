```python
cols_to_use = ['Suburb','Rooms','Method','Type','Regionname','Propertycount','Distance', 'Postcode', 'Bedroom2', 'Bathroom', 'Car',
       'Landsize','BuildingArea','Price']
df = df[cols_to_use]
df.head()
```

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
# Fill the null or missing value 
cols_to_fill_zero=['Propertycount','Distance','Bedroom2','Bathroom','Car']
df[cols_to_fill_zero] = df[cols_to_fill_zero].fillna(0)
df.isna().sum()
```

```python
df['Landsize'] = df['Landsize'].fillna(df.Landsize.mean())
```

```python
df.isnull().sum()
```

```python
df['BuildingArea'] = df['BuildingArea'].fillna(df.BuildingArea.mean())
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
ch_mean = df.loc[df["BuildingArea"] !=0,'BuildingArea'].mean()
ch_mean
```

```python
# Calculate the of these given feature
ch_mean = df.loc[df['Landsize'] !=0,'Landsize'].mean()
ch_mean
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df = pd.get_dummies(df,drop_first = True)
df.head()
```

```python
df.head()
```

```python
x = df.drop('Price',axis=1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
train_x, test_x , train_y , test_y = train_test_split(x, y, test_size=0.3 , random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression().fit(train_x , train_y)
```

```python
reg.score(test_x , test_y)
```

```python
reg.score(train_x,train_y)
```
