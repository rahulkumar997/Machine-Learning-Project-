```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X,y_true = make_blobs(n_samples = 500 , centers = 3 , cluster_std = 0.60,random_state=42)
```

```python
X
```

```python
df = pd.DataFrame(X,columns=['Feature_1','Feature_2'])
```

```python
df
```

```python
df.head()
```

```python
df.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# WCSS(Within clustring sum of square distance) and it is also called as the Inertia
wcss = []
K_range = range(1,11)

```

```python
wcss = []
K_range = range(1, 11)  # for example

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(K_range , wcss, marker = 'o')
# These are the Elbow Methode
```

```python
kmeans_final = KMeans(n_clusters=3 , random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue = df['cluster'],
                palette='viridis')
```

```python
kmean_final1 = KMeans(n_clusters=6 , random_state=42)
```

```python
cluster_labels1 = kmean_final1.fit_predict(X_scaled)
```

```python
cluster_labels1
```

```python
df['cluster'] = cluster_labels1
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue = df['cluster'],
                palette='viridis')
```

```python
plt.plot(K_range, wcss, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.grid(True)
plt.show()

```
